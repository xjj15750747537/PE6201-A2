**简体中文：** 只修改标有 `EDIT ONLY THIS LINE` 的值。先核对模型官方 OpenRouter 定价页；价格是公开 metadata，不是 secret。在小组同意使用 credits 前保持 `RUN_FULL_BATTERY = False`；仅在最终 55-case / 93-trial run 时改为 `True`：每个 booking 案例运行一次，每个 negative 案例运行三次。


```python
# EDIT ONLY THIS LINE: use a unique public label, e.g. hing-gpt4omini-v2
RUN_NAME = 'EDIT-unique-run-name'
# EDIT ONLY THIS LINE: public name or initials only; never a key
OWNER = 'EDIT-owner'
# EDIT ONLY THIS LINE: exact OpenRouter model identifier
MODEL = 'EDIT-model-id'
# EDIT ONLY THIS LINE: model family, e.g. OpenAI, Anthropic, Google
FAMILY = 'EDIT-model-family'
# EDIT ONLY THIS LINE: cheap, mid, or frontier
PRICE_TIER = 'cheap'
# EDIT ONLY THIS LINE: verified public US$ input price per 1M tokens
INPUT_PRICE_PER_MILLION = 0.0
# EDIT ONLY THIS LINE: verified public US$ output price per 1M tokens
OUTPUT_PRICE_PER_MILLION = 0.0
# EDIT ONLY THIS LINE: v2 for the final battery; v1 only for one same-model comparison
PROMPT_VERSION = 'v2'
# EDIT ONLY THIS LINE: set True only after team approval to spend credits
RUN_FULL_BATTERY = False

import subprocess

if not RUN_FULL_BATTERY:
    print('Safe preflight only: no API call was made. Complete the public settings, obtain team approval, then set RUN_FULL_BATTERY = True.')
else:
    public_values = [RUN_NAME, OWNER, MODEL, FAMILY]
    if any(value.startswith('EDIT-') for value in public_values):
        raise ValueError('Replace every EDIT- value before starting a paid live run.')
    if PRICE_TIER not in {'cheap', 'mid', 'frontier'} or PROMPT_VERSION not in {'v1', 'v2'}:
        raise ValueError('Check PRICE_TIER and PROMPT_VERSION.')
    if INPUT_PRICE_PER_MILLION <= 0 or OUTPUT_PRICE_PER_MILLION <= 0:
        raise ValueError('Use verified positive public token prices; do not guess.')
    command = [
        'python3', 'scripts/run_live_battery.py',
        '--run-name', RUN_NAME, '--owner', OWNER, '--model', MODEL, '--family', FAMILY,
        '--price-tier', PRICE_TIER, '--prompt-version', PROMPT_VERSION,
        '--input-price-per-million', str(INPUT_PRICE_PER_MILLION),
        '--output-price-per-million', str(OUTPUT_PRICE_PER_MILLION),
    ]
    subprocess.run(command, cwd=REPOSITORY, check=True)
    print('Measured D5(b) fragment saved. Do not edit files in results/live_runs/.')
```
